```python
source_path = f'{landing_folder_path}/constructors.json'
table_name = f'{catalog_name}.{bronze_schema}.constructors'
```

```python
constructors_schema = 'constructorId STRING, name STRING, nationality STRING, url STRING' ## DDL String
```

```python
constructors_df = (
    spark.read
        .format('json')
        .schema(constructors_schema)
        .load(source_path)
)
```

```python
display(constructors_df)
```

```python
constructors_df_final = add_ingestion_metadata(constructors_df)
```

```python
display(constructors_df_final)
```

```python
(
    constructors_df_final.write
        .format('delta')
        .mode('overwrite')
        .saveAsTable(table_name)
)
```
